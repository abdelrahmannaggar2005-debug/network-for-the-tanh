```python
import numpy as np

np.random.seed(42)

W1 = np.random.uniform(-0.5, 0.5, size=(2, 2))
W2 = np.random.uniform(-0.5, 0.5, size=(2, 2)) #output weight 

b1 = 0.5
b2 = 0.7

x = np.array([0.8, -0.2])

net1 = x @ W1 + b1
out1 = np.tanh(net1)

net2 = out1 @ W2 + b2
output = np.tanh(net2)

print("Input")
print(x)
print("Weights W1")
print(np.round(W1, 4))
print("Weights W2")
print(np.round(W2, 4))
print("Hidden layer net net1")
print(np.round(net1, 4))
print("Hidden layer out1")
print(np.round(out1, 4))
print("Output layer net2")
print(np.round(net2, 4))
print("Final output")
print(np.round(output, 4))
```
